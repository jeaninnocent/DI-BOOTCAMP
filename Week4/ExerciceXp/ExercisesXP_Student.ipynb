```python

# This piece of code is already prefilled, run it to execute it and see the results.
# It demonstrates standard classification metrics for binary loan default prediction.

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, average_precision_score, confusion_matrix, classification_report

# Please replace these placeholders with your true labels and predicted probabilities.
y_true = [0,1,0,1,0,0,1,0,1,0]            # placeholder labels
y_pred_proba = [0.05,0.80,0.10,0.65,0.20,0.15,0.70,0.30,0.85,0.25]  # placeholder probabilities

# You should set a decision threshold that reflects the precision–recall trade-off for your business case.
threshold = 0.5
y_pred = [1 if p >= threshold else 0 for p in y_pred_proba]

print("Accuracy:", round(accuracy_score(y_true, y_pred), 4))
print("Precision:", round(precision_score(y_true, y_pred, zero_division=0), 4))
print("Recall:", round(recall_score(y_true, y_pred, zero_division=0), 4))
print("F1-score:", round(f1_score(y_true, y_pred, zero_division=0), 4))
print("ROC-AUC:", round(roc_auc_score(y_true, y_pred_proba), 4))
print("PR-AUC (Average Precision):", round(average_precision_score(y_true, y_pred_proba), 4))
print("\nConfusion matrix:\n", confusion_matrix(y_true, y_pred))
print("\nClassification report:\n", classification_report(y_true, y_pred, zero_division=0))
```

Pour évaluer un modèle d'apprentissage supervisé, tel qu'un algorithme de classification conçu pour anticiper des crises médicales à partir de données biométriques, la stratégie repose sur la comparaison rigoureuse entre les prédictions et les étiquettes réelles à l'aide d'une validation croisée (cross-validation). Les performances sont mesurées via des métriques clés tirées de la matrice de confusion : la précision (pour limiter les fausses alertes), le rappel (absolument crucial ici pour ne manquer aucune urgence ou défaut réel) et le score F1 qui équilibre les deux, tout en s'appuyant sur l'aire sous la courbe ROC (AUC) pour évaluer la capacité globale de séparation des classes. Le défi principal réside dans la gestion des données déséquilibrées (les anomalies étant souvent rares par rapport aux données normales) et dans le risque de surapprentissage, ce qui exige une optimisation minutieuse pour garantir que le modèle se généralise bien face à de nouvelles situations réelles.

Dans le cadre de l'apprentissage non supervisé, comme l'utilisation d'un algorithme de clustering pour analyser et regrouper les trajets complexes des usagers des transports urbains afin d'optimiser les lignes locales, l'évaluation est plus délicate car il n'y a pas de vérité terrain préétablie pour vérifier les réponses. La stratégie consiste à mesurer la cohérence mathématique et géométrique des groupes formés en utilisant des métriques internes telles que le score de silhouette, qui évalue à la fois la cohésion au sein d'un même groupe et la séparation par rapport aux autres groupes, ou encore la méthode du coude pour justifier le nombre optimal de clusters. La limite majeure de cette approche est son aspect hautement subjectif : une segmentation mathématiquement parfaite n'a de valeur que si elle correspond à une réalité métier exploitable sur le terrain, ce qui exige systématiquement une interprétation humaine approfondie pour valider la pertinence des résultats.

L'évaluation d'un modèle d'apprentissage par renforcement nécessite quant à elle une approche épisodique, puisque l'agent apprend de manière dynamique en interagissant avec son environnement pour maximiser un signal de retour. La performance est généralement mesurée en figeant l'apprentissage pour analyser la récompense cumulative moyenne obtenue sur de nombreux épisodes de test indépendants, tout en observant la vitesse de convergence du modèle vers une politique optimale stable. L'enjeu critique lors de cette évaluation est de vérifier que le modèle maintient un bon équilibre entre l'exploration et l'exploitation, le principal défi étant la conception même de la fonction de récompense (reward shaping) afin d'éviter que l'agent ne trouve des failles dans l'environnement pour accumuler des points sans résoudre le problème de fond.
