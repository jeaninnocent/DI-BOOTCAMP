```python

# This piece of code is already prefilled, run it to execute it and see the results.
# It demonstrates standard classification metrics for binary loan default prediction.

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, average_precision_score, confusion_matrix, classification_report

# Please replace these placeholders with your true labels and predicted probabilities.
y_true = [0,1,0,1,0,0,1,0,1,0]            # placeholder labels
y_pred_proba = [0.05,0.80,0.10,0.65,0.20,0.15,0.70,0.30,0.85,0.25]  # placeholder probabilities

# You should set a decision threshold that reflects the precision–recall trade-off for your business case.
threshold = 0.5
y_pred = [1 if p >= threshold else 0 for p in y_pred_proba]

print("Accuracy:", round(accuracy_score(y_true, y_pred), 4))
print("Precision:", round(precision_score(y_true, y_pred, zero_division=0), 4))
print("Recall:", round(recall_score(y_true, y_pred, zero_division=0), 4))
print("F1-score:", round(f1_score(y_true, y_pred, zero_division=0), 4))
print("ROC-AUC:", round(roc_auc_score(y_true, y_pred_proba), 4))
print("PR-AUC (Average Precision):", round(average_precision_score(y_true, y_pred_proba), 4))
print("\nConfusion matrix:\n", confusion_matrix(y_true, y_pred))
print("\nClassification report:\n", classification_report(y_true, y_pred, zero_division=0))
```

1. Choix des modèles
Pour ce type de problème tabulaire de classification binaire, plusieurs approches sont recommandées :

Régression Logistique : Excellent modèle de base (baseline). Il est rapide, interprétable et permet de comprendre le poids de chaque caractéristique.

Forêts Aléatoires (Random Forest) ou Gradient Boosting (XGBoost / LightGBM) : Ces modèles ensemblistes sont extrêmement performants sur les données tabulaires, gèrent bien les relations non linéaires et sont robustes face aux valeurs aberrantes.

2. Étapes d'évaluation et Métriques

Séparation des données : Diviser le jeu de données en ensembles d'entraînement (80%) et de test (20%).

Traitement du déséquilibre : Les cas de défauts sont généralement minoritaires. Il faudra peut-être utiliser des techniques comme SMOTE (Synthetic Minority Over-sampling Technique) ou ajuster les poids des classes.

Métriques clés :

Le Rappel (Recall / Sensibilité) : C'est la métrique la plus cruciale ici. Il mesure la capacité du modèle à identifier tous les vrais défauts de paiement. Dans le domaine bancaire, un faux négatif (accorder un prêt à quelqu'un qui va faire défaut) coûte beaucoup plus cher qu'un faux positif (refuser un prêt à un bon payeur).

Score F1 : Pour maintenir un bon équilibre entre la Précision et le Rappel.

AUC-ROC (Area Under the Receiver Operating Characteristic Curve) : Pour évaluer la capacité globale du modèle à distinguer les deux classes, indépendamment du seuil de classification choisi.
