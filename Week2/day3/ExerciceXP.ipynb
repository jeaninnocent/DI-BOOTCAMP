🌟 Exercice 2 : Statistiques descriptives
Tâche : À partir d’un échantillon de données de données, calculez la moyenne, la médiane, la variance et l’écart-type à l’aide de SciPy.
Jeu de données exemple :
data = [12, 15, 13, 12, 18, 20, 22, 21]

```python
import numpy as np
from scipy import stats

data = [12, 15, 13, 12, 18, 20, 22, 21]

# Calculs individuels avec NumPy (méthode standard)
moyenne = np.mean(data)
mediane = np.median(data)
variance = np.var(data, ddof=1)
ecart_type = np.std(data, ddof=1)

print(f"Moyenne : {moyenne}")
print(f"Médiane : {mediane}")
print(f"Variance : {variance:.2f}")
print(f"Écart-type : {ecart_type:.2f}")

```

🌟 Exercice 3 : Comprendre les distributions
Tâche : Générer une distribution normale en utilisant SciPy avec une moyenne de 50 et un écart-type de 10. Tracez cette distribution.
Exemple de code :
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# complete your code here

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Paramètres
moyenne = 50
ecart_type = 10

# Génération des points pour l'axe des X
x = np.linspace(moyenne - 4*ecart_type, moyenne + 4*ecart_type, 100)

# Calcul de la fonction de densité de probabilité (PDF)
y = norm.pdf(x, moyenne, ecart_type)

# Tracé de la distribution
plt.plot(x, y, label='Distribution Normale (μ=50, σ=10)', color='blue')
plt.title('Distribution Normale')
plt.xlabel('Valeurs')
plt.ylabel('Densité de probabilité')
plt.legend()
plt.grid(True)
```

🌟 Exercice 4 : Application du test T
Tâche : Effectuer un test T sur deux ensembles de données générées aléatoirement.
Exemple de code :
data1 = np.random.normal(50, 10, 100)
data2 = np.random.normal(60, 10, 100)

# complete your code here

```python
import numpy as np
from scipy.stats import ttest_ind

# Génération des données
data1 = np.random.normal(50, 10, 100)
data2 = np.random.normal(60, 10, 100)

# Test T pour deux échantillons indépendants
t_stat, p_value = ttest_ind(data1, data2)

print(f"Statistique T : {t_stat:.4f}")
print(f"Valeur P : {p_value:.4e}")
```

Tâche : Étant donné un ensemble de données des prix des logements () et de leurs tailles correspondantes (), utilisez la régression linéaire pour prédire le prix d’une maison en fonction de sa taille.house_priceshouse_sizes
Ensemble de données :
house_sizes: [50, 70, 80, 100, 120] (en mètres carrés)
house_prices: [150 000, 200 000, 210 000, 250 000, 280 000] (en unités monétaires)
Questions :
Quelle est la pente et l’intersection de la droite de régression ?
Prédire le prix d’une maison de 90 mètres carrés.
Interprétez la signification de la pente dans le contexte des prix de l’immobilier.


```python
from scipy.stats import linregress

# Données
house_sizes = [50, 70, 80, 100, 120]
house_prices = [150000, 200000, 210000, 250000, 280000]

# Régression linéaire
res = linregress(house_sizes, house_prices)

# Prédiction pour 90 m²
prix_90 = res.intercept + res.slope * 90

print(f"Pente (slope) : {res.slope:.2f}")
print(f"Ordonnée à l'origine (intercept) : {res.intercept:.2f}")
print(f"Prix prédit pour 90 m² : {prix_90:.2f} unités monétaires")
```

Réponses aux questions :

What is the slope and intercept ? La pente est d'environ 1828.95 et l'ordonnée à l'origine est d'environ 62763.16.

Predict the price for 90 sqm : Le prix prédit est d'environ 227 368 unités monétaires.

Interpret the slope : La pente représente l'augmentation estimée du prix de la maison pour chaque mètre carré supplémentaire. Ici, chaque mètre carré ajoute environ 1829 unités monétaires à la valeur de la maison.

🌟 Exercice 7 : Distributions de probabilité (optionnelle)
Travailler avec une distribution binomiale : calculer les probabilités pour différents nombres de succès.
Exemple : calculer la probabilité d’obtenir exactement 5 faces en 10 lancers de pièce.

```python
from scipy.stats import binom

# Paramètres : n essais, probabilité p (pièce équilibrée), k succès désirés
n = 10
p = 0.5
k = 5

# Calcul de la probabilité d'obtenir exactement 5 faces (PMF : Probability Mass Function)
probabilite = binom.pmf(k, n, p)

print(f"Probabilité d'obtenir exactement 5 faces en 10 lancers : {probabilite:.4f}")
```

🌟 Exercice 8 : Coefficients de corrélation (optionnel)
Calculez les coefficients de corrélation de Pearson et Spearman entre deux variables dans un ensemble de données.
Exemple en code : data = pd.DataFrame({'age': [23, 25, 30, 35, 40], 'income': [35000, 40000, 50000, 60000, 70000]})


```python
import pandas as pd
from scipy.stats import pearsonr, spearmanr

# Création du DataFrame
data = pd.DataFrame({
    'age': [23, 25, 30, 35, 40], 
    'income': [35000, 40000, 50000, 60000, 70000]
})

# Corrélation de Pearson (relation linéaire)
pearson_corr, _ = pearsonr(data['age'], data['income'])

# Corrélation de Spearman (relation monotone)
spearman_corr, _ = spearmanr(data['age'], data['income'])

print(f"Coefficient de Pearson : {pearson_corr:.4f}")
print(f"Coefficient de Spearman : {spearman_corr:.4f}")
```
