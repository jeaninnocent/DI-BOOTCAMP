```python
# Calcul de la factorielle avec une fonction récursive
def factorielle(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorielle(n - 1)

# Exemple d'utilisation
nombre = 5
print(f"La factorielle de {nombre} est {factorielle(nombre)}")
```
