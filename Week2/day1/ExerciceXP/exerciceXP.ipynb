Qu’est-ce que l’analyse de données ?
L'analyse de données est le processus d'inspection, de nettoyage, de transformation et de modélisation des données dans le but de découvrir des informations utiles, de formuler des conclusions et de soutenir la prise de décision.

Pourquoi l’analyse des données est-elle importante dans les contextes modernes ?
Dans un monde saturé d'informations, l'analyse permet aux organisations de ne plus naviguer à l'aveugle. Elle permet de remplacer les suppositions par des faits concrets, d'optimiser les performances, de comprendre les comportements des utilisateurs et de prévoir les tendances futures.

Trois domaines d'application concrets :

La Santé (Modélisation prédictive) : L'analyse des données de santé (via des objets connectés) permet de développer des modèles capables de prédire des urgences médicales imminentes, comme l'anticipation des crises d'AVC à l'aide de capteurs biométriques.

Transport et Mobilité Urbaine : Les données sont utilisées pour digitaliser les réseaux de transport urbain (comme les bus ou les minicars locaux). L'analyse des flux permet de gérer les billetteries en ligne, de prévoir les heures de départ et d'optimiser les trajets en temps réel.

Cybersécurité : L'analyse continue des logs et du trafic réseau permet d'identifier des comportements anormaux. Les algorithmes détectent les menaces en temps réel, garantissant ainsi une "Sécurité by Design" dans le développement d'applications.

Exercice 9 : Exportez une dataframe en format Excel et JSON.
Créez un dataframe simple.
Exportez la dataframe dans un fichier Excel.
Exportez la dataframe en fichier JSON.


```python
# 1. Création d'un DataFrame simple
data = {
    'Nom': ['Alice', 'Bob', 'Charlie'],
    'Âge': [25, 30, 35],
    'Langage': ['Python', 'JavaScript', 'C++']
}
df_simple = pd.DataFrame(data)

# 2. Exportation vers Excel (nécessite openpyxl: pip install openpyxl)
df_simple.to_excel('mon_dataframe.xlsx', index=False)
print("Fichier Excel exporté avec succès.")

# 3. Exportation vers JSON
df_simple.to_json('mon_dataframe.json', orient='records', force_ascii=False)
print("Fichier JSON exporté avec succès.")
```
