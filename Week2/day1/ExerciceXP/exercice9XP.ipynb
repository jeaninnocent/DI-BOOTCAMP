```python
import pandas as pd
```

```python
# 1. Création d'un dataframe simple
donnees = {
    'Nom': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'Profession': ['Ingénieur', 'Médecin', 'Analyste']
}
```

```python
df_simple = pd.DataFrame(donnees)
```

```python
# 2. Exportation en fichier Excel (nécessite d'avoir installé 'openpyxl' : pip install openpyxl)
df_simple.to_excel('mon_dataframe.xlsx', index=False)
```

```python
# 3. Exportation en fichier JSON
df_simple.to_json('mon_dataframe.json', orient='records')

print("Fichiers exportés avec succès !")
```
