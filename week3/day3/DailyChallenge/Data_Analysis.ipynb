1. Chargement et exploration des données
Commençons par importer les bibliothèques nécessaires, charger le jeu de données et observer sa structure initiale.

```python
# Importation des bibliothèques requises
import pandas as pd
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt

# 1. Chargement des données 
df = pd.read_csv('train.csv')

# 2. Exploration de la structure initiale
print("Aperçu des 5 premières lignes :")
display(df.head())

print("\nInformations sur le jeu de données (Types de données et valeurs non nulles) :")
df.info()

print("\nStatistiques descriptives de base :")
display(df.describe())
```

L'utilisation de info() nous permet de confirmer que la plupart (ou la totalité) des caractéristiques de ce jeu de données sont déjà de type numérique (int64 ou float64). La variable cible est price_range (qui va généralement de 0 à 3).

2. Nettoyage des données et prétraitement
Bien que ce jeu de données soit généralement propre, il est crucial de vérifier les valeurs manquantes et de s'assurer que les données sont prêtes pour l'analyse.

```python
# Vérification des valeurs manquantes
missing_values = df.isnull().sum()
print("Valeurs manquantes par colonne :\n", missing_values[missing_values > 0])

# Traitement des valeurs manquantes (si elles existent)
if df.isnull().values.any():
    print("\nDes valeurs manquantes ont été trouvées. Remplacement par la médiane...")
    df.fillna(df.median(), inplace=True)
else:
    print("\nAucune valeur manquante détectée. Le dataset est propre.")

# Les données catégorielles (comme le Bluetooth 'blue', le 'dual_sim', etc.) 
# sont déjà encodées en 0 et 1 dans ce dataset spécifique.
```

Si des valeurs nulles avaient été présentes, nous aurions utilisé l'imputation par la médiane pour éviter d'être biaisés par des valeurs aberrantes (outliers). L'encodage binaire natif facilite la transition directe vers l'analyse statistique.

3. Analyse statistique avec NumPy et SciPy
Nous allons examiner la distribution de nos données et effectuer des tests d'hypothèses pour voir si certaines caractéristiques (comme la RAM) ont une différence statistiquement significative en fonction de la gamme de prix.

```python
# Sélection d'une caractéristique clé pour l'analyse détaillée : 'ram' (Mémoire vive)
ram_data = df['ram'].values

# 3.1 Mesures de tendance centrale et de variabilité avec NumPy
mean_ram = np.mean(ram_data)
median_ram = np.median(ram_data)
mode_ram = stats.mode(ram_data, keepdims=True)[0][0]
range_ram = np.ptp(ram_data) # peak-to-peak (max - min)
var_ram = np.var(ram_data, ddof=1)
std_ram = np.std(ram_data, ddof=1)

print(f"--- Analyse de la RAM ---")
print(f"Moyenne: {mean_ram:.2f}, Médiane: {median_ram:.2f}, Mode: {mode_ram}")
print(f"Plage: {range_ram}, Variance: {var_ram:.2f}, Écart-type: {std_ram:.2f}")

# 3.2 Évaluation des formes de distribution avec SciPy (Asymétrie et Kurtose)
skewness = stats.skew(ram_data)
kurtosis = stats.kurtosis(ram_data)
print(f"Asymétrie (Skewness): {skewness:.2f}")
print(f"Kurtose: {kurtosis:.2f}")
# Une asymétrie proche de 0 indique une distribution relativement symétrique.

# 3.3 Test d'hypothèse (ANOVA) : La RAM diffère-t-elle selon la gamme de prix ?
# Nous divisons la RAM en groupes basés sur 'price_range'
groups = [df['ram'][df['price_range'] == i] for i in df['price_range'].unique()]
f_stat, p_value = stats.f_oneway(*groups)

print(f"\n--- Test ANOVA sur la RAM par rapport à la gamme de prix ---")
print(f"Statistique F : {f_stat:.2f}, P-value : {p_value:.2e}")
if p_value < 0.05:
    print("Conclusion : Il y a une différence statistiquement significative de la RAM entre les différentes gammes de prix.")

# 3.4 Corrélations caractéristiques-cibles à l'aide de SciPy
corr_coef, p_val_corr = stats.pearsonr(df['ram'], df['price_range'])
print(f"\nCorrélation Pearson entre RAM et Prix : {corr_coef:.2f} (P-value: {p_val_corr:.2e})")
```

La P-value extrêmement faible du test ANOVA confirme que la RAM n'est pas distribuée au hasard entre les catégories de prix ; elle est un facteur discriminant majeur.

4. Visualisation des données avec Matplotlib
Nous utilisons Matplotlib pour créer des visualisations claires qui appuient nos découvertes statistiques.

```python
# Configuration globale de Matplotlib
plt.rcParams['figure.figsize'] = (10, 6)

# 4.1 Histogramme : Distribution de la puissance de la batterie
plt.figure()
plt.hist(df['battery_power'], bins=20, color='skyblue', edgecolor='black')
plt.title('Distribution de la puissance de la batterie')
plt.xlabel('Puissance de la batterie (mAh)')
plt.ylabel('Fréquence')
plt.grid(axis='y', alpha=0.75)
plt.show()

# 4.2 Nuage de points : RAM vs Puissance de la batterie, coloré par la gamme de prix
plt.figure()
scatter = plt.scatter(df['ram'], df['battery_power'], c=df['price_range'], cmap='viridis', alpha=0.6)
plt.title('Relation entre la RAM, la Batterie et la Gamme de Prix')
plt.xlabel('RAM (Mo)')
plt.ylabel('Batterie (mAh)')
plt.colorbar(scatter, label='Gamme de prix (0 à 3)')
plt.show()

# 4.3 Boîte à moustaches (Boxplot) : RAM par gamme de prix
plt.figure()
plt.boxplot(groups, labels=df['price_range'].unique())
plt.title('Distribution de la RAM pour chaque Gamme de Prix')
plt.xlabel('Gamme de Prix')
plt.ylabel('RAM (Mo)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()

# 4.4 Carte thermique (Heatmap) de corrélation
plt.figure(figsize=(12, 10))
corr_matrix = df.corr()
plt.imshow(corr_matrix, cmap='coolwarm', interpolation='none', aspect='auto')
plt.colorbar()
plt.xticks(range(len(corr_matrix)), corr_matrix.columns, rotation=90)
plt.yticks(range(len(corr_matrix)), corr_matrix.columns)
plt.title('Matrice de Corrélation des Caractéristiques')
plt.show()
```

Le nuage de points montre une stratification claire : les téléphones avec beaucoup de RAM se situent systématiquement dans la gamme de prix la plus élevée.

La boîte à moustaches confirme l'absence de chevauchement significatif de la RAM entre les téléphones bon marché (0) et les téléphones phares (3).

La carte thermique met en évidence une très forte corrélation positive (couleur rouge foncé) entre ram et price_range. Les caractéristiques comme la résolution de l'écran (px_width, px_height) et la batterie (battery_power) montrent également des corrélations positives modérées.

5. Synthèse de l’intuition et conclusion
Conclusions principales de l'analyse :

Le déterminant principal (La RAM) : Les tests statistiques (corrélation de Pearson très élevée, ANOVA avec p-value proche de zéro) et les visualisations (boxplots sans chevauchement majeur) prouvent que la Mémoire Vive (RAM) est le facteur numéro un pour prédire et justifier le prix d'un téléphone mobile dans ce jeu de données.

Les déterminants secondaires : La carte thermique de corrélation révèle que la capacité de la batterie (battery_power) et la résolution de l'écran (px_height, px_width) jouent également un rôle significatif, bien que moindre comparé à la RAM, dans l'augmentation du prix.

Découverte inattendue : Des fonctionnalités matérielles souvent marketées comme premium (comme la 3G/4G, le Bluetooth, ou le nombre de cœurs du processeur n_cores) montrent une corrélation étonnamment faible avec la gamme de prix cible. Cela suggère que dans ce marché spécifique, les consommateurs ou les fabricants valorisent avant tout les performances brutes (RAM/Batterie) plutôt que la connectivité de base qui est devenue une norme sur presque toutes les gammes.

Action exploitable : Si une entreprise souhaite fabriquer un téléphone abordable (gamme 0 ou 1), elle peut intégrer la 4G et un bon processeur sans faire exploser les coûts, mais elle devra faire des compromis stricts sur la quantité de RAM intégrée.
