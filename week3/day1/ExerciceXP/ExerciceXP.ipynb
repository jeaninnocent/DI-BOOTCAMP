```python
import numpy as np
```

```python
# Créer un tableau 1D de 0 à 9
array_1d = np.arange(10)
print(array_1d)
```

```python
# Convertir une liste en tableau NumPy avec le type entier
liste = [3.14, 2.17, 0, 1, 2]
array_int = np.array(liste, dtype=int)
print(array_int)
```

```python
# Créer un tableau 3x3 avec des valeurs de 1 à 9
array_3x3 = np.arange(1, 10).reshape(3, 3)
print(array_3x3)
```

```python
# Créer un tableau 2D (4, 5) de nombres aléatoires arrondis à 2 décimales
array_random = np.round(np.random.rand(4, 5), 2)
print(array_random)
```

```python
# Sélectionner la deuxième ligne
array_2d = np.array([[21, 22, 23, 22, 22],
                     [20, 21, 22, 23, 24],
                     [21, 22, 23, 22, 22]])

second_row = array_2d[1]
print(second_row)
```

```python
# Inverser l'ordre des éléments
array_to_reverse = np.arange(10)
reversed_array = array_to_reverse[::-1]
print(reversed_array)
```

```python
# Créer une matrice d'identité 4x4
identity_matrix = np.eye(4)
print(identity_matrix)
```

```python
# Trouver la somme et la moyenne
array_agg = np.arange(10) # Utilisation du tableau de l'exercice 1
somme = np.sum(array_agg)
moyenne = np.mean(array_agg)

print(f"Sum: {somme}, Average: {moyenne}")
```

```python
# Créer de 1 à 20 et redimensionner en 4x5
array_reshaped = np.arange(1, 21).reshape(4, 5)
print(array_reshaped)
```

```python
# Extraire les nombres impairs
array_base = np.arange(10)
odd_numbers = array_base[array_base % 2 != 0]
print(odd_numbers)
```
