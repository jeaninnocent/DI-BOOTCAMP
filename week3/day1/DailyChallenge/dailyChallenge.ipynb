```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

1. Data Preparation:

Hint 1: Use np.random.uniform(low, high, size) to generate the temperature data.
Hint 2: Create a DataFrame using pd.DataFrame(data, index, columns) with appropriate index and columns.

Use NumPy to generate a synthetic dataset representing average monthly temperatures (in degrees Celsius) for 12 months across 10 different cities. The temperatures should range from -5 to 35 degrees.

Convert this NumPy array into a Pandas DataFrame, adding city names as index and months as columns.

```python
# Fixer la graine aléatoire pour obtenir les mêmes résultats à chaque exécution
np.random.seed(42)

# --- 1. PRÉPARATION DES DONNÉES ---
villes = ['Abidjan', 'Montréal', 'Paris', 'Dakar', 'Tokyo', 
          'New York', 'Moscou', 'Sydney', 'Le Caire', 'Rio']
mois = ['Jan', 'Fév', 'Mar', 'Avr', 'Mai', 'Juin', 
        'Juil', 'Août', 'Sep', 'Oct', 'Nov', 'Déc']

# Génération des températures entre -5 et 35 degrés
temperatures = np.random.uniform(low=-5.0, high=35.0, size=(10, 12))

# Création du DataFrame
df_meteo = pd.DataFrame(temperatures, index=villes, columns=mois)

print("Aperçu des données générées :")
display(df_meteo.head())

# --- 2. ANALYSE DES DONNÉES ---
# Calcul de la température moyenne annuelle (sur l'axe des colonnes, donc axis=1)
moyennes_annuelles = df_meteo.mean(axis=1)

# Identification des extrêmes
ville_plus_chaude = moyennes_annuelles.idxmax()
ville_plus_froide = moyennes_annuelles.idxmin()

temp_max = moyennes_annuelles.max()
temp_min = moyennes_annuelles.min()

print(f"\nVille la plus chaude : {ville_plus_chaude} ({temp_max:.2f}°C)")
print(f"Ville la plus froide : {ville_plus_froide} ({temp_min:.2f}°C)")
```

2. Data Analysis:

Hint 1: Calculate the annual average temperature using DataFrame.mean(axis).
Hint 2: Find the city with the highest and lowest average temperature using idxmax() and idxmin() methods.

Calculate the annual average temperature for each city.

Identify the city with the highest and lowest average temperature for the year.
3. Data Visualization:

```python
# Configuration du style global
plt.style.use('seaborn-v0_8-darkgrid')
fig = plt.figure(figsize=(16, 12))

# --- Graphique 1 : Évolution des températures mensuelles ---
ax1 = fig.add_subplot(2, 1, 1)
for ville in df_meteo.index:
    # Mettre en évidence les villes extrêmes
    if ville == ville_plus_chaude:
        ax1.plot(df_meteo.columns, df_meteo.loc[ville], label=f"{ville} (Max)", linewidth=3, color='red')
    elif ville == ville_plus_froide:
        ax1.plot(df_meteo.columns, df_meteo.loc[ville], label=f"{ville} (Min)", linewidth=3, color='blue')
    else:
        ax1.plot(df_meteo.columns, df_meteo.loc[ville], label=ville, linewidth=1, alpha=0.5)

ax1.set_title("Évolution des températures mensuelles par ville", fontsize=16, fontweight='bold')
ax1.set_xlabel("Mois", fontsize=12)
ax1.set_ylabel("Température (°C)", fontsize=12)
ax1.legend(loc='upper right', bbox_to_anchor=(1.15, 1))

# --- Graphique 2 : Comparaison des moyennes annuelles ---
ax2 = fig.add_subplot(2, 1, 2)
couleurs = ['skyblue' if v != ville_plus_chaude and v != ville_plus_froide else 
            ('red' if v == ville_plus_chaude else 'blue') for v in moyennes_annuelles.index]

moyennes_annuelles.sort_values().plot(kind='bar', color=couleurs, ax=ax2)

ax2.set_title("Température moyenne annuelle par ville", fontsize=16, fontweight='bold')
ax2.set_xlabel("Villes", fontsize=12)
ax2.set_ylabel("Température Moyenne (°C)", fontsize=12)
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```
