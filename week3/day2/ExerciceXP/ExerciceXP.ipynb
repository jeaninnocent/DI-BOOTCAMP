🌟 Exercise 1 : Matrix Operations
Instructions
In this exercise, you’ll work with a 3x3 matrix. Here’s a brief explanation of the concepts:

Determinant: The determinant is a value that can be computed from the elements of a square matrix. It provides important information about the matrix, such as whether it has an inverse, and is used in various areas like linear algebra and calculus. To understand more about it you can watch this video.
Inverse of a Matrix: The inverse of a matrix is a matrix that, when multiplied with the original matrix, results in an identity matrix. Not all matrices have inverses. The inverse is crucial in solving systems of linear equations.
Create a 3x3 matrix and perform the following operations:

Calculate the determinant.
Find the inverse of the matrix.


```python
# Création d'une matrice 3x3
matrix = np.array([
    [2, 1, 1],
    [1, 3, 2],
    [1, 0, 0]
])

# 1. Calcul du déterminant
det = np.linalg.det(matrix)
print(f"Déterminant de la matrice : {det:.2f}")

# 2. Calcul de l'inverse (possible uniquement si le déterminant != 0)
if det != 0:
    matrix_inv = np.linalg.inv(matrix)
    print("\nInverse de la matrice :")
    print(matrix_inv)
else:
    print("La matrice n'est pas inversible (déterminant = 0).")
```

 Exercise 2 : Statistical Analysis
Instructions
In this exercise, you’ll calculate statistical measures for a dataset:

Mean: The average value of a dataset.
Median: The middle value in a dataset when it is arranged in ascending or descending order.
Standard Deviation: A measure of the amount of variation or dispersion in a set of values.
Using NumPy, generate an array of 50 random numbers and compute:

The mean and median.
The standard deviation.

```python
# Génération d'un tableau de 50 nombres aléatoires (distribution normale)
data = np.random.randn(50) * 100 

# Calcul des mesures statistiques
mean_val = np.mean(data)
median_val = np.median(data)
# Utilisation de ddof=0 pour la population complète, ou ddof=1 pour un échantillon
std_val = np.std(data, ddof=1) 

print(f"Moyenne : {mean_val:.2f}")
print(f"Médiane : {median_val:.2f}")
print(f"Écart-type : {std_val:.2f}")
```

🌟 Exercise 6 : Basic Hypothesis Testing
Instructions
Create a sample dataset to test the effectiveness of a new training program on employee productivity:


import numpy as np

# Productivity scores of employees before the training program
productivity_before = np.random.normal(loc=50, scale=10, size=30)

# Productivity scores of the same employees after the training program
productivity_after = productivity_before + np.random.normal(loc=5, scale=3, size=30)

# Your task is to formulate a hypothesis regarding the training program's effectiveness 
# and test it using basic statistical functions in NumPy.


Given a dataset, formulate a simple hypothesis and test it using basic statistical functions in NumPy.



Formulation de l'hypothèse :Hypothèse Nulle ($H_0$) : Le programme de formation n'a eu aucun effet, ou a diminué la productivité (Différence moyenne $\le$ 0).Hypothèse Alternative ($H_1$) : Le programme de formation a augmenté la productivité (Différence moyenne > 0).

```python
# Données de l'exercice
productivity_before = np.random.normal(loc=50, scale=10, size=30)
productivity_after = productivity_before + np.random.normal(loc=5, scale=3, size=30)

# Calcul des différences de productivité pour chaque employé
differences = productivity_after - productivity_before

# Tests statistiques de base avec NumPy
mean_diff = np.mean(differences)
std_diff = np.std(differences, ddof=1) # ddof=1 car on travaille sur un échantillon
n = len(differences)

# Calcul d'un score T simplifié pour voir si la différence est significative
t_stat = mean_diff / (std_diff / np.sqrt(n))

print(f"Amélioration moyenne de la productivité : {mean_diff:.2f} points")
print(f"Statistique T : {t_stat:.2f}")

if t_stat > 1.699: # Valeur critique approximative pour alpha=0.05 et df=29 (unilatéral)
    print("Conclusion : On rejette H0. Le programme d'entraînement a été efficace.")
else:
    print("Conclusion : On ne peut pas rejeter H0. L'efficacité n'est pas prouvée.")
```

🌟 Exercise 8 : Time Series Data Manipulation
Instructions
Generate time series data for the year 2023. Demonstrate slicing for the following intervals:

January to March
April to June
July to September
October to December
Generate a time series data for a specific period and demonstrate how to slice this data for different intervals.

```python
# Génération de la série temporelle pour toute l'année 2023
year_2023 = np.arange('2023-01-01', '2024-01-01', dtype='datetime64[D]')

# Découpage (slicing) par intervalles en utilisant des masques booléens
q1 = year_2023[(year_2023 >= np.datetime64('2023-01-01')) & (year_2023 < np.datetime64('2023-04-01'))]
q2 = year_2023[(year_2023 >= np.datetime64('2023-04-01')) & (year_2023 < np.datetime64('2023-07-01'))]
q3 = year_2023[(year_2023 >= np.datetime64('2023-07-01')) & (year_2023 < np.datetime64('2023-10-01'))]
q4 = year_2023[(year_2023 >= np.datetime64('2023-10-01')) & (year_2023 < np.datetime64('2024-01-01'))]

print(f"Jours de Janvier à Mars : {len(q1)} jours")
print(f"Jours d'Avril à Juin : {len(q2)} jours")
print(f"Premier jour de Juillet à Septembre : {q3[0]}")
print(f"Dernier jour d'Octobre à Décembre : {q4[-1]}")
```

 Exercise 9 : Data Conversion
Instructions
Demonstrate how to convert a NumPy array to a Pandas DataFrame and vice versa.



```python
# 1. De NumPy vers Pandas DataFrame
np_matrix = np.array([
    [1, 2, 3],
    [4, 5, 6]
])
df_converted = pd.DataFrame(np_matrix, columns=['Col_A', 'Col_B', 'Col_C'])

print("--- De NumPy à Pandas ---")
print(df_converted)
print(type(df_converted))

# 2. De Pandas DataFrame vers NumPy Array
print("\n--- De Pandas à NumPy ---")
# .to_numpy() est la méthode recommandée (préférable à .values)
back_to_numpy = df_converted.to_numpy() 

print(back_to_numpy)
print(type(back_to_numpy))
```

🌟 Exercise 10 : Basic Visualization
Instructions
Use Matplotlib to visualize a simple dataset created with NumPy (e.g., a line graph of random numbers).





```python
# Génération des données avec NumPy
# 100 points espacés uniformément entre 0 et 10
x = np.linspace(0, 10, 100) 
# Calcul du sinus de x + ajout d'un petit bruit aléatoire pour rendre ça plus réaliste
y = np.sin(x) + np.random.normal(0, 0.1, 100)

# Création du graphique avec Matplotlib
plt.figure(figsize=(10, 5)) # Définition de la taille de la figure
plt.plot(x, y, label='Signal bruité', color='b', marker='o', markersize=3, linestyle='-')

# Ajout des détails
plt.title('Visualisation d\'une onde sinusoïdale avec bruit')
plt.xlabel('Axe X')
plt.ylabel('Axe Y')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)

# Affichage du graphique
plt.show()
```
