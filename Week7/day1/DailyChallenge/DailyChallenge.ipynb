Solving the frequency problem using TF-IDF
When we create a BoW out from some text, all the words are treated equaly as importants. Like “Alice” or “say” in a fantasy book about Alice. We expected those words to be repeated all over the book, making them not so informative to us.

The solution for this problem would be to consider the frequency relative to the corpus. In this case, if there is a word in a document that doesn’t appears much in the other documents, it is likely meaningful and should be considered more important. And the same way in the oposite: A word that is repeated a lot in all the documents will be considered less important.

That’s the perfect situation to use TF-IDF (Term Frequency-Inverse Document Frequency)

```python
import re
import requests
import nltk
import spacy
import matplotlib.pyplot as plt
from wordcloud import WordCloud
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from nltk.stem import PorterStemmer
from nltk.tokenize import word_tokenize

# Téléchargements NLTK obligatoires
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('averaged_perceptron_tagger')
nltk.download('maxent_ne_chunker')
nltk.download('words')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('maxent_ne_chunker_tab') # Added to resolve LookupError

# Chargement du modèle spaCy
nlp = spacy.load('en_core_web_sm')

# URLs d'exemple du Projet Gutenberg pour Lewis Carroll
urls = [
    "https://www.gutenberg.org/files/11/11-0.txt",       # Alice's Adventures in Wonderland
    "https://www.gutenberg.org/files/12/12-0.txt",       # Through the Looking-Glass
    "https://www.gutenberg.org/cache/epub/29042/pg29042.txt" # A Tangled Tale
]
book_titles = ["Alice in Wonderland", "Through the Looking-Glass", "A Tangled Tale"]

# ==========================================
# TEXT PREPROCESSING
# ==========================================

### 1 & 2. Chargement des textes et nettoyage (Slicing Gutenberg)
def load_texts(urls_list):
    corpus = []
    for url in urls_list:
        response = requests.get(url)
        response.encoding = 'utf-8'
        text = response.text

        # Slicing pour enlever les licences de début et de fin du Projet Gutenberg
        start_idx = text.find("*** START OF")
        if start_idx == -1: start_idx = text.find("START")

        end_idx = text.find("*** END OF")
        if end_idx == -1: end_idx = text.find("END")

        if start_idx != -1 and end_idx != -1:
            text = text[start_idx:end_idx]

        # Nettoyage des caractères spéciaux non-mots (garder lettres et espaces)
        cleaned_text = re.sub(r'[^a-zA-Z\s]', '', text)
        # Normalisation des espaces superflus
        cleaned_text = re.sub(r'\s+', ' ', cleaned_text).strip()
        corpus.append(cleaned_text)
    return corpus

print("--- 1 & 2. Chargement et Slicing des textes ---")
raw_corpus = load_texts(urls)

for i, text in enumerate(raw_corpus):
    print(f"\n[Aperçu 200 caractères] - {book_titles[i]}:\n", text[:200], "...")

### 3. Tokenisation
print("\n--- 3. Tokenisation (150 premiers tokens) ---")
tokenized_corpus = [word_tokenize(text.lower()) for text in raw_corpus]
for i, tokens in enumerate(tokenized_corpus):
    print(f"\nTokens {book_titles[i]}:\n", tokens[:150])

### 4. Suppression des Stopwords
print("\n--- 4. Suppression et Vérification des Stopwords ---")
stop_words = set(nltk.corpus.stopwords.words('english'))
filtered_corpus_tokens = [[w for w in tokens if w not in stop_words] for tokens in tokenized_corpus]

# Vérification empirique sur "i" et "me"
for i, tokens in enumerate(filtered_corpus_tokens):
    count_i = tokens.count('i')
    count_me = tokens.count('me')
    print(f"{book_titles[i]} : Occurrences de 'i' -> {count_i}, 'me' -> {count_me} (0 attendu)")

### 5. Racinisation (Stemming)
print("\n--- 5. Stemming (PorterStemmer) ---")
ps = PorterStemmer()
stemmed_corpus_tokens = [[ps.stem(w) for w in tokens] for tokens in filtered_corpus_tokens]
for i, tokens in enumerate(stemmed_corpus_tokens):
    print(f"50 premiers stems - {book_titles[i]}:\n", tokens[:50])

### 6. Lemmatisation (spaCy)
print("\n--- 6. Lemmatisation (spaCy sur texte brut filtré) ---")
lemmatized_corpus = []
for i, text in enumerate(raw_corpus):
    # Traitement spaCy par morceaux pour éviter de saturer la mémoire si le livre est massif
    doc = nlp(text[:50000]) # On limite à l'échantillon de début pour la démo
    lemmas = [token.lemma_.lower() for token in doc if not token.is_stop and token.is_alpha]
    lemmatized_corpus.append(lemmas)
    print(f"50 premiers lemmes - {book_titles[i]}:\n", lemmas[:50])

### 7. Analyse de la différence Stemming vs Lemmatisation
"""
Analyse :
- Le Stemming (Porter) utilise des règles heuristiques brutales pour couper les fins de mots
  (ex: 'running' -> 'run', 'studies' -> 'studi'). Il génère parfois des non-mots.
- La Lemmatisation (spaCy) s'appuie sur un dictionnaire et une analyse morphologique complète.
  Elle retourne la forme du dictionnaire (le lemme), garantissant un vrai mot de la langue
  (ex: 'saw' devient 'see' ou 'saw' selon le contexte grammatical).
"""

### 8 & 9. POS Tagging et Reconnaissance d'Entités Nommées (NER)
print("\n--- 8 & 9. POS Tagging & NER ---")
for i, tokens in enumerate(tokenized_corpus):
    pos_tags = nltk.pos_tag(tokens[:100]) # Échantillon de 100 mots
    ner_chunks = nltk.ne_chunk(pos_tags[:30])
    print(f"\nPOS Tags (10 premiers) - {book_titles[i]} :\n", pos_tags[:10])
    print(f"Entités détectées (Arbre NER condensé) - {book_titles[i]} :\n", [child for child in ner_chunks if isinstance(child, nltk.Tree)][:3])

# ==========================================
# ANALYSING THE TEXT (VISUALIZATIONS)
# ==========================================

### 1. Génération des Word Clouds
print("\n--- Génération des Word Clouds... ---")
for i, tokens in enumerate(filtered_corpus_tokens):
    text_for_cloud = " ".join(tokens)
    wordcloud = WordCloud(width=800, height=400, background_color='white').generate(text_for_cloud)

    plt.figure(figsize=(10, 5))
    plt.imshow(wordcloud, interpolation='bilinear')
    plt.title(f"Word Cloud - {book_titles[i]}", fontsize=16)
    plt.axis('off')
    plt.show()

### 2 & 3. Bag of Words (BoW) & Fréquences
print("\n--- 2 & 3. Analyse Bag of Words (BoW) ---")
# Le meilleur texte à utiliser est le texte LEMMATISÉ pour éviter les doublons sémantiques.
joined_lemmas = [" ".join(lemmas) for lemmas in lemmatized_corpus]

vectorizer = CountVectorizer()
bow_matrix = vectorizer.fit_transform(joined_lemmas)
word_features = vectorizer.get_feature_names_out()

# Extraction des mots les plus fréquents cumulés
sum_words = bow_matrix.sum(axis=0)
words_freq = [(word, sum_words[0, idx]) for word, idx in vectorizer.vocabulary_.items()]
words_freq = sorted(words_freq, key=lambda x: x[1], reverse=True)

print("Top 5 mots les plus fréquents (BoW global) :", words_freq[:5])

# Exemple d'identification des numéros requis à la question 3 :
print(f"\nStructure de la matrice BoW au format (Document, Index du mot) Fréquence :")
print(str(bow_matrix[:1])) # Affiche les coordonnées du premier document

### 4 & 5. Diagramme Circulaire (BoW) & Analyse
top5_bow = words_freq[:5]
labels_bow = [item[0] for item in top5_bow]
sizes_bow = [item[1] for item in top5_bow]

plt.figure(figsize=(6, 6))
plt.pie(sizes_bow, labels=labels_bow, autopct='%1.1f%%', startangle=140)
plt.title("Top 5 Mots les plus fréquents (Méthode Bag of Words)")
plt.show()

"""
Analyse BoW : Les mots extraits (ex: "alice", "say", "little") sont très attendus.
Ils décrivent l'univers ou la structure des phrases narratives mais ne capturent pas
la spécificité discriminante de chaque livre par rapport aux autres.
"""

# ==========================================
# SOLVING THE FREQUENCY PROBLEM USING TF-IDF
# ==========================================

### 1 & 2. TF-IDF Vectorizer & Pie Plots individuels
print("\n--- Résolution de la fréquence via TF-IDF ---")
tfidf_vec = TfidfVectorizer(min_df=1, max_df=2)
tfidf_matrix = tfidf_vec.fit_transform(joined_lemmas)
tfidf_features = tfidf_vec.get_feature_names_out()

# Création des graphiques en secteurs pour chaque livre selon le score TF-IDF
for doc_idx, title in enumerate(book_titles):
    row = tfidf_matrix.getrow(doc_idx).toarray()[0]
    top_indices = row.argsort()[-5:][::-1]

    labels_tfidf = [tfidf_features[i] for i in top_indices]
    values_tfidf = [row[i] for i in top_indices]

    # Éviter le plantage si un document n'a pas assez de termes filtrés
    if sum(values_tfidf) == 0:
        continue

    plt.figure(figsize=(5, 5))
    plt.pie(values_tfidf, labels=labels_tfidf, autopct='%1.1f%%', startangle=140)
    plt.title(f"Top 5 termes discriminants (TF-IDF) -\n {title}")
    plt.show()
```
